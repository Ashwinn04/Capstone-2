# Load & inspect dataset

```python
# Step 1: load_and_inspect.py
import pandas as pd
pd.set_option("display.max_columns", 200)

df = pd.read_csv("D:\Research Paper\engineered_data.csv")   # change path if needed
print("Shape:", df.shape)
print("\nColumns (first 60):", df.columns.tolist()[:60])
print("\nTarget column 'SepsisLabel' present?:", 'SepsisLabel' in df.columns)
print("\nSepsisLabel distribution:\n", df['SepsisLabel'].value_counts(dropna=False))
print("\nSample rows:")
display(df.head(8))
```

# Compute clinical scores (SIRS, qSOFA, NEWS2, Partial SOFA)

```python
# Step 2: compute_scores.py
import numpy as np

def compute_sirs(row):
    cnt = 0
    t = row.get('Temp', np.nan)
    if pd.notna(t) and (t > 38.0 or t < 36.0): cnt += 1
    hr = row.get('HR', np.nan)
    if pd.notna(hr) and hr > 90: cnt += 1
    rr = row.get('Resp', np.nan)
    if pd.notna(rr) and rr > 20: cnt += 1
    wbc = row.get('WBC', np.nan)
    if pd.notna(wbc) and (wbc > 12 or wbc < 4): cnt += 1
    return cnt

def compute_qsofa(row):
    # prefer dataset column if available
    if 'qSOFA_Score_simple' in row.index:
        v = row.get('qSOFA_Score_simple', np.nan)
        if pd.notna(v): return int(v)
    # fallback: Resp >=22, SBP <=100; no GCS available
    rr = row.get('Resp', np.nan)
    sbp = row.get('SBP', np.nan)
    score = 0
    if pd.notna(rr) and rr >= 22: score += 1
    if pd.notna(sbp) and sbp <= 100: score += 1
    return score

def compute_news2(row):
    score = 0
    rr = row.get('Resp', np.nan)
    if pd.notna(rr):
        if rr <= 8: score += 3
        elif 9 <= rr <= 11: score += 1
        elif 12 <= rr <= 20: score += 0
        elif 21 <= rr <= 24: score += 2
        else: score += 3
    spo2 = row.get('O2Sat', row.get('SaO2', np.nan))
    if pd.notna(spo2):
        if spo2 <= 91: score += 3
        elif 92 <= spo2 <= 93: score += 2
        elif 94 <= spo2 <= 95: score += 1
    sbp = row.get('SBP', np.nan)
    if pd.notna(sbp):
        if sbp <= 90: score += 3
        elif 91 <= sbp <= 100: score += 2
        elif 101 <= sbp <= 110: score += 1
    hr = row.get('HR', np.nan)
    if pd.notna(hr):
        if hr <= 40: score += 3
        elif 41 <= hr <= 50: score += 1
        elif 51 <= hr <= 90: score += 0
        elif 91 <= hr <= 110: score += 1
        elif 111 <= hr <= 130: score += 2
        else: score += 3
    temp = row.get('Temp', np.nan)
    if pd.notna(temp):
        if temp <= 35.0: score += 3
        elif 35.1 <= temp <= 36.0: score += 1
        elif 36.1 <= temp <= 38.0: score += 0
        elif 38.1 <= temp <= 39.0: score += 1
        else: score += 2
    return score

def compute_sofa_partial(row):
    score = 0
    # Platelets
    p = row.get('Platelets', np.nan)
    if pd.notna(p):
        if p >= 150: score += 0
        elif 100 <= p < 150: score += 1
        elif 50 <= p < 100: score += 2
        elif 20 <= p < 50: score += 3
        else: score += 4
    # Bilirubin_total
    b = row.get('Bilirubin_total', np.nan)
    if pd.notna(b):
        if b < 1.2: score += 0
        elif b < 2.0: score += 1
        elif b < 6.0: score += 2
        elif b < 12.0: score += 3
        else: score += 4
    # Creatinine
    cr = row.get('Creatinine', np.nan)
    if pd.notna(cr):
        if cr < 1.2: score += 0
        elif cr < 2.0: score += 1
        elif cr < 3.5: score += 2
        elif cr < 5.0: score += 3
        else: score += 4
    # MAP crude cardio surrogate
    mapv = row.get('MAP', np.nan)
    if pd.notna(mapv):
        if mapv >= 70: score += 0
        elif 50 <= mapv < 70: score += 1
        else: score += 2
    return score

# apply to dataframe (assumes df loaded)
df['SIRS_count'] = df.apply(compute_sirs, axis=1)
df['SIRS_flag'] = (df['SIRS_count'] >= 2).astype(int)
df['qSOFA'] = df.apply(compute_qsofa, axis=1)
df['NEWS2'] = df.apply(compute_news2, axis=1)
df['SOFA_partial'] = df.apply(compute_sofa_partial, axis=1)

print("Added SIRS_count, SIRS_flag, qSOFA, NEWS2, SOFA_partial")
display(df[['SIRS_count','SIRS_flag','qSOFA','NEWS2','SOFA_partial']].head(8))
```

# Feature engineering & preprocessing

```python
# Step 3: preprocess.py
from sklearn.impute import SimpleImputer

candidate_features = [
    'HR','O2Sat','Temp','SBP','MAP','DBP','Resp','FiO2','pH','PaCO2','SaO2',
    'BaseExcess','HCO3','WBC','Platelets','Creatinine','Bilirubin_total','Age',
    'SIRS_count','SIRS_flag','qSOFA','NEWS2','SOFA_partial'
]
features = [f for f in candidate_features if f in df.columns]
print("Features used:", features)

# create X, y (optionally sample to speed up)
X = df[features].copy()
y = df['SepsisLabel'].astype(int)

# median impute
imp = SimpleImputer(strategy='median')
X_imputed = pd.DataFrame(imp.fit_transform(X), columns=features)

# optional scaling (for LR)
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = pd.DataFrame(scaler.fit_transform(X_imputed), columns=features)

# Save prepared data for next steps
X_imputed.to_csv("X_imputed.csv", index=False)
X_scaled.to_csv("X_scaled.csv", index=False)
print("Saved X_imputed.csv and X_scaled.csv")
```

# Train/test split (patient-level leakage prevention)

```python
# Step 4: split_data.py
from sklearn.model_selection import train_test_split

# If Patient_ID exists, split by patient (prevents leakage across time/hours)
if 'Patient_ID' in df.columns:
    # split patient IDs into train/test, then filter rows
    unique_pats = df['Patient_ID'].unique()
    train_pats, test_pats = train_test_split(unique_pats, test_size=0.2, random_state=42)
    train_idx = df['Patient_ID'].isin(train_pats)
    test_idx  = df['Patient_ID'].isin(test_pats)
    X_train = X_imputed[train_idx]
    y_train = y[train_idx]
    X_test = X_imputed[test_idx]
    y_test = y[test_idx]
    print("Patient-level split done. Train patients:", len(train_pats), "Test patients:", len(test_pats))
else:
    # standard stratified split
    X_train, X_test, y_train, y_test = train_test_split(X_imputed, y, stratify=y, test_size=0.2, random_state=42)
    print("Random stratified split done. Train size:", X_train.shape, "Test size:", X_test.shape)
```

# rain baseline models (Logistic Regression, Random Forest, XGBoost/HGB fallback)

```python
# Step 5: train_models.py
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, roc_auc_score, classification_report
from sklearn.ensemble import HistGradientBoostingClassifier
import joblib

# scale features for LR
scaler_lr = StandardScaler()
X_train_scaled = scaler_lr.fit_transform(X_train)
X_test_scaled  = scaler_lr.transform(X_test)

# Logistic Regression
lr = LogisticRegression(max_iter=1000, class_weight='balanced', solver='liblinear', random_state=42)
lr.fit(X_train_scaled, y_train)
y_pred_lr = lr.predict(X_test_scaled)
y_proba_lr = lr.predict_proba(X_test_scaled)[:,1]

# Random Forest
rf = RandomForestClassifier(n_estimators=100, class_weight='balanced', random_state=42, n_jobs=-1)
rf.fit(X_train, y_train)
y_pred_rf = rf.predict(X_test)
y_proba_rf = rf.predict_proba(X_test)[:,1]

# XGBoost if installed else HistGradientBoosting fallback
try:
    import xgboost as xgb
    xgb_clf = xgb.XGBClassifier(use_label_encoder=False, eval_metric='logloss', n_estimators=100, random_state=42)
    xgb_clf.fit(X_train, y_train)
    y_pred_xgb = xgb_clf.predict(X_test)
    y_proba_xgb = xgb_clf.predict_proba(X_test)[:,1]
    xgb_used = True
except Exception:
    xgb_clf = HistGradientBoostingClassifier(random_state=42)
    xgb_clf.fit(X_train, y_train)
    y_pred_xgb = xgb_clf.predict(X_test)
    try:
        y_proba_xgb = xgb_clf.predict_proba(X_test)[:,1]
    except:
        y_proba_xgb = y_pred_xgb
    xgb_used = False

# Save models and scalers
joblib.dump(lr, "model_logistic.pkl")
joblib.dump(rf, "model_rf.pkl")
joblib.dump(xgb_clf, "model_xgb_or_hgb.pkl")
joblib.dump(scaler_lr, "scaler_lr.pkl")
print("Saved models: model_logistic.pkl, model_rf.pkl, model_xgb_or_hgb.pkl and scaler_lr.pkl")
```

# Evaluation (metrics & feature importance)

```python
# Step 6: evaluate.py
def eval_model(name, y_true, y_pred, y_proba):
    print("=== %s ===" % name)
    print("Accuracy:", round(accuracy_score(y_true, y_pred),4))
    print("Precision:", round(precision_score(y_true, y_pred, zero_division=0),4))
    print("Recall:", round(recall_score(y_true, y_pred),4))
    try:
        print("ROC AUC:", round(roc_auc_score(y_true, y_proba),4))
    except:
        print("ROC AUC: N/A (probabilities not available)")
    print(classification_report(y_true, y_pred, digits=4))

eval_model("Logistic Regression", y_test, y_pred_lr, y_proba_lr)
eval_model("Random Forest", y_test, y_pred_rf, y_proba_rf)
eval_model("XGBoost/HGB", y_test, y_pred_xgb, y_proba_xgb)

# Feature importances
import pandas as pd
if hasattr(rf, "feature_importances_"):
    importances = pd.Series(rf.feature_importances_, index=X_train.columns).sort_values(ascending=False)
    print("\nTop RF importances:\n", importances.head(15))

if hasattr(lr, "coef_"):
    coefs = pd.Series(lr.coef_.ravel(), index=X_train.columns)
    print("\nTop LR coefficient magnitudes:\n", coefs.abs().sort_values(ascending=False).head(15))
```

# Save outputs + inference function

```python
# Step 7: save_and_infer.py
import joblib, json
import numpy as np

# Load models/scaler (if new session)
lr = joblib.load("model_logistic.pkl")
rf = joblib.load("model_rf.pkl")
xgb_clf = joblib.load("model_xgb_or_hgb.pkl")
scaler_lr = joblib.load("scaler_lr.pkl")

# Save test predictions with probabilities
out = X_test.copy()
out['SepsisLabel_true'] = y_test.values
out['pred_lr'] = y_pred_lr
out['prob_lr'] = y_proba_lr
out['pred_rf'] = y_pred_rf
out['prob_rf'] = y_proba_rf
out['pred_xgb'] = y_pred_xgb
try:
    out['prob_xgb'] = y_proba_xgb
except:
    out['prob_xgb'] = np.nan
out.to_csv("test_predictions_with_scores.csv", index=False)
print("Saved test_predictions_with_scores.csv")

# Inference helper: accepts a single-row dict or pandas Series and returns computed scores + predictions
def infer_single(input_row):
    # input_row: dict or pandas Series with same feature names
    row = pd.Series(input_row)
    # compute clinical scores (reuse functions)
    row['SIRS_count'] = compute_sirs(row)
    row['SIRS_flag'] = int(row['SIRS_count'] >= 2)
    row['qSOFA'] = compute_qsofa(row)
    row['NEWS2'] = compute_news2(row)
    row['SOFA_partial'] = compute_sofa_partial(row)
    # build feature vector
    feat_vec = pd.DataFrame([row[features].fillna(pd.Series(X_imputed.median()))])
    # LR needs scaling
    feat_scaled = scaler_lr.transform(feat_vec)
    prob_lr = lr.predict_proba(feat_scaled)[:,1][0]
    prob_rf = rf.predict_proba(feat_vec)[0,1]
    try:
        prob_xg = xgb_clf.predict_proba(feat_vec)[0,1]
    except:
        prob_xg = xgb_clf.predict(feat_vec)[0]
    return {
        'SIRS_count': int(row['SIRS_count']),
        'qSOFA': int(row['qSOFA']),
        'NEWS2': int(row['NEWS2']),
        'SOFA_partial': int(row['SOFA_partial']),
        'prob_lr': float(prob_lr),
        'prob_rf': float(prob_rf),
        'prob_xgb': float(prob_xg)
    }

# Example usage:
# new_patient = {'HR': 95, 'Resp': 24, 'SBP': 98, 'Temp': 39, 'WBC': 14, 'O2Sat': 88, 'Platelets': 140, 'Creatinine': 1.5, 'Bilirubin_total': 1.0, 'Age': 65, ...}
# print(infer_single(new_patient))
```
